```python

import random

# ==================== KNOWLEDGE BASE (updated March 2026) ====================
INTENTS = {
    "greeting": {
        "patterns": ["hi", "hello", "hey", "sup", "yo", "morning", "evening"],
        "responses": [
            "Hello, Red Devils fan! Glory Glory Man United! How can I help you today?",
            "Hey there! Ready to talk Manchester United? What's on your mind?"
        ]
    },
    "goodbye": {
        "patterns": ["bye", "goodbye", "exit", "quit", "see you", "later"],
        "responses": [
            "Goodbye! Come back anytime for more Man United chat. Glory Glory!",
            "See you later, mate! Keep the red flag flying high!"
        ]
    },
    "history": {
        "patterns": ["history", "founded", "origin", "past", "munich", "busby", "sir matt"],
        "responses": [
            "Manchester United was founded in 1878 as Newton Heath LYR F.C. and renamed in 1902. Key moments: first league title in 1908, Munich air disaster (1958), and first European Cup win in 1968 under Sir Matt Busby. The club has an incredible legacy!"
        ]
    },
    "manager": {
        "patterns": ["manager", "coach", "head coach", "carrick", "who manages", "current boss"],
        "responses": [
            "The current head coach (until end of 2025/26 season) is Michael Carrick. He was appointed in January 2026 after Ruben Amorim's departure."
        ]
    },
    "captain": {
        "patterns": ["captain", "who is captain", "bruno fernandes"],
        "responses": [
            "The captain is Bruno Fernandes — our Portuguese midfield maestro and leader!"
        ]
    },
    "squad": {
        "patterns": ["squad", "team", "players", "current players", "first team"],
        "responses": [
            "Key players in the 2025-26 squad include:\n• Bruno Fernandes (Captain, MF)\n• Marcus Rashford (FW)\n• Kobbie Mainoo (MF)\n• André Onana (GK)\n• Harry Maguire (DF)\n• Casemiro (MF)\n• Diogo Dalot (DF)\n• Alejandro Garnacho (FW)\n• Lisandro Martínez (DF)\n• Amad Diallo (FW)\nAnd many more rising stars!"
        ]
    },
    "trophies": {
        "patterns": ["trophies", "honours", "titles", "won", "champions league", "premier league"],
        "responses": [
            "Manchester United have won:\n• 20 English top-flight titles\n• 13 FA Cups\n• 6 League Cups\n• 3 UEFA Champions League titles\n• 1 Europa League\nAnd many more! The most successful club in English football history."
        ]
    },
    "stadium": {
        "patterns": ["stadium", "old trafford", "theatre of dreams", "home ground"],
        "responses": [
            "Home is Old Trafford — 'The Theatre of Dreams' — with a capacity of 74,244. It's one of the most iconic stadiums in the world!"
        ]
    },
    "nickname": {
        "patterns": ["nickname", "red devils", "what are they called"],
        "responses": [
            "We are the Red Devils! Club colours: red shirts, white shorts, black socks."
        ]
    },
    "rivals": {
        "patterns": ["rival", "derby", "man city", "liverpool", "arsenal"],
        "responses": [
            "Biggest rivals: Manchester City (Manchester derby), Liverpool, and Arsenal. Nothing beats beating those lot!"
        ]
    },
    "news": {
        "patterns": ["news", "latest", "what's happening", "update"],
        "responses": [
            "For the absolute latest news and fixtures, visit the official site: https://www.manutd.com\nRight now Michael Carrick is in charge and the team is pushing hard!"
        ]
    },
    "fallback": {
        "patterns": [],
        "responses": [
            "Sorry, I didn't quite catch that. Try asking about the history, current squad, manager, trophies, Old Trafford, or just say 'tell me about the team'.",
            "Hmm, not sure about that one. Want info on Bruno Fernandes, the squad, trophies, or anything else Man United related?"
        ]
    }
}

# ==================== CHATBOT LOGIC ====================
def get_response(user_input):
    user_input = user_input.lower().strip()

    # Check each intent
    for intent_name, data in INTENTS.items():
        if intent_name == "fallback":
            continue
        for pattern in data["patterns"]:
            if pattern in user_input:
                return random.choice(data["responses"])

    # If nothing matched
    return random.choice(INTENTS["fallback"]["responses"])


# ==================== MAIN CONVERSATION LOOP ====================
print("=" * 60)
print("🏟️  MANCHESTER UNITED CHATBOT  🏟️")
print("Ask me anything about the Red Devils!")
print("Type 'bye' to exit.")
print("=" * 60)

while True:
    user_input = input("\nYou: ")

    if user_input.lower() in ["bye", "goodbye", "exit", "quit", "see you"]:
        print(get_response(user_input))
        break

    response = get_response(user_input)
    print(f"ManUtdBot: {response}")
```
